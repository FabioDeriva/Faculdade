```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split

tf.random.set_seed(42)
np.random.seed(42)
```

# <font color='green'><u><b>Parte 1 - Conjunto de Dados</b></u></font>

Não altere o conteúdo da célula a seguir!

```python
# Carregando o dataset CIFAR-10
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Ajustando dimensões dos rótulos
y_train = y_train.reshape(-1,)
y_test = y_test.reshape(-1,)

# Separando validação do treino
x_train, x_val, y_train, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

# Classes do dataset
class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]
```

O CIFAR-10 é um dataset usado em classificação de imagens, criado como base de referência para treinar e testar algoritmos de aprendizado de máquina. Ele possui 60000 imagens coloridas de baixa resolução (32x32), divididas em 10 classes com 6000 imagens em cada uma, contendo coisas como animais (gato, cachorro, pássaro, sapo, cavalo, cervo) e meios de transporte (avião, carro, navio, caminhão).

```python
print(x_train.shape)
print(x_val.shape)
print(x_test.shape)
print(x_train[0].shape)
print(y_train.shape)
print(y_val.shape)
print(y_test.shape)
print(np.unique(y_train))
print(len(np.unique(y_train)))
```

Existem 40000 imagens para treino, 10000 para validação e 10000 para teste. Cada imagem tem resolução 32x32 e 3 canais (RGB), e o dataset possui 10 classes diferentes, numeradas de 0 a 9.

### <font color='#2D9CDB'>Q5) Verifique os valores mínimo e máximo dos pixels das imagens do conjunto de treinamento. Em seguida, normalize os conjuntos de treino, validação e teste dividindo os valores dos pixels por 255. # <font color='green'><u><b>Parte 2 - MLP</b></u></font>

### <font color='#2D9CDB'>Q7) As redes neurais do tipo Multilayer Perceptron (MLP) utilizam camadas densamente conectadas e, portanto, esperam vetores unidimensionais como entrada. Explique por que é necessário transformar as imagens do CIFAR-10 antes de utilizá-las em uma MLP e discuta o que acontece com a estrutura espacial das imagens durante esse processo. Em seguida, construa uma rede neural MLP utilizando a API Sequential do Keras com uma camada Flatten, duas camadas densas escondidas com função de ativação ReLU e uma camada de saída com 10 neurônios e ativação Softmax.</font>

```python
modelo_mlp = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(32, 32, 3)),

    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dense(128, activation='relu'),

    tf.keras.layers.Dense(10, activation='softmax')
])

modelo_mlp.summary()
```

A MLP só trabalha com vetores, então as imagens que estão no formato (32, 32, 3) precisam virar vetores unidimensionais antes de entrar na rede. A camada Flatten faz essa transformação, juntando todos os pixels em um vetor de 3072 valores. Durante esse processo a estrutura espacial da imagem é perdida, ou seja, a rede não sabe mais quais pixels eram vizinhos na imagem original.

### <font color='#2D9CDB'>Q8) Exiba o resumo da arquitetura da rede neural criada utilizando o método summary() e analise a quantidade de parâmetros treináveis da MLP. Com base nas dimensões das imagens do CIFAR-10 e no número de neurônios utilizados nas camadas densas, discuta por que MLPs podem rapidamente se tornar modelos muito grandes para problemas de classificação de imagens.</font>

```python
modelo_mlp.summary()
```

A MLP tem mais de 800 mil parâmetros, sendo a maior parte concentrada na primeira camada densa. Isso acontece porque o Flatten transforma a imagem em um vetor de 3072 valores e cada um precisa estar conectado a todos os 256 neurônios da camada seguinte. Conforme a resolução aumenta, a quantidade de parâmetros cresce muito rápido, o que torna a MLP pesada e pouco eficiente para imagens.

### <font color='#2D9CDB'>Q9) Compile a MLP utilizando o otimizador Adam com taxa de aprendizado de $10^{-3}$, a função de custo sparse_categorical_crossentropy e a métrica accuracy. Em seguida, explique o papel da função de custo durante o treinamento, o objetivo do algoritmo de otimização e por que a combinação entre Softmax e entropia cruzada é adequada para problemas de classificação multiclasse.</font>

```python
modelo_mlp.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

A função de custo mede o quanto as previsões do modelo estão erradas em relação às classes corretas, e o objetivo do treinamento é reduzir esse valor. O otimizador Adam ajusta os pesos da rede a cada época para minimizar a função de custo de forma eficiente. A combinação entre Softmax e entropia cruzada é boa para classificação multiclasse porque a Softmax transforma a saída em probabilidades para cada uma das 10 classes, enquanto a sparse_categorical_crossentropy compara essas probabilidades com o rótulo certo e penaliza erros. Como os rótulos do CIFAR-10 estão em formato numérico (0 a 9), a versão sparse é a apropriada.

### <font color='#2D9CDB'>Q10) Treine a rede neural por 25 épocas utilizando os conjuntos de treinamento e validação. Armazene o histórico do treinamento em um objeto do tipo History e, em seguida, plote gráficos mostrando a evolução da loss e da accuracy ao longo das épocas para os conjuntos de treinamento e validação. Após gerar os gráficos, analise o comportamento do treinamento, discutindo se o modelo aparenta convergir adequadamente, se há estabilidade durante o treinamento e se existem indícios de overfitting.</font>

```python
historico_mlp = modelo_mlp.fit(
    x_train,
    y_train,
    epochs=25,
    validation_data=(x_val, y_val)
)

plt.figure(figsize=(8, 5))
plt.plot(historico_mlp.history['loss'], label='Treino')
plt.plot(historico_mlp.history['val_loss'], label='Validação')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(historico_mlp.history['accuracy'], label='Treino')
plt.plot(historico_mlp.history['val_accuracy'], label='Validação')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

Durante o treinamento dá pra ver que a loss de treino tende a diminuir e a accuracy de treino aumenta, mostrando que a MLP está aprendendo. Mas se a accuracy de validação para de melhorar ou a loss de validação começa a subir, isso indica overfitting. Como a MLP usa as imagens achatadas em vetores, ela perde parte da estrutura espacial e tem mais dificuldade para generalizar.

### <font color='#2D9CDB'>Q11) Avalie o desempenho da MLP utilizando o conjunto de teste e reporte os valores de loss e accuracy obtidos. Em seguida, compare os resultados dos conjuntos de treinamento e teste e discuta se o modelo aparenta apresentar boa capacidade de generalização.</font>

```python
train_loss, train_accuracy = modelo_mlp.evaluate(x_train, y_train)
test_loss, test_accuracy = modelo_mlp.evaluate(x_test, y_test)

print(train_loss)
print(train_accuracy)

print(test_loss)
print(test_accuracy)
```

A accuracy no teste fica em torno de 50% com loss próxima de 1,4. Se a accuracy de teste estiver bem próxima da de treino, o modelo generaliza bem, mas se ficar muito menor isso indica overfitting. Como a MLP perde a estrutura espacial dos pixels ao usar Flatten, o desempenho no CIFAR-10 fica limitado comparado a uma CNN.

### <font color='#2D9CDB'>Q12) Utilize a MLP treinada para gerar previsões sobre algumas imagens do conjunto de teste. Exiba as probabilidades previstas para cada classe, identifique a classe predita pelo modelo e compare-a com a classe verdadeira da imagem. Com base nas probabilidades obtidas, explique o significado da saída Softmax e discuta se o modelo está realizando previsões com alta ou baixa confiança.</font>

```python
num_imagens = 5

previsoes = modelo_mlp.predict(x_test[:num_imagens])

plt.figure(figsize=(12, 4))

for i in range(num_imagens):
    classe_predita = np.argmax(previsoes[i])
    classe_verdadeira = y_test[i]

    plt.subplot(1, num_imagens, i + 1)
    plt.imshow(x_test[i])
    plt.title(
        "Pred: " + class_names[classe_predita] +
        "\nReal: " + class_names[classe_verdadeira]
    )
    plt.axis('off')

plt.show()

for i in range(num_imagens):
    print("Imagem", i, "-> probabilidades:", previsoes[i])
    print("Confiança:", np.max(previsoes[i]))
    print()
```

A saída Softmax mostra uma distribuição de probabilidades entre as 10 classes, e a soma dessas probabilidades é aproximadamente 1. Cada valor indica o quanto o modelo acredita que a imagem pertence àquela classe. Pela capacidade limitada da MLP, as previsões aparentam confiança apenas satisfatória, com algumas probabilidades baixas e algumas classes próximas, mostrando que o modelo fica em dúvida em vários casos.

### <font color='#2D9CDB'>Q13) Plote a matriz de confusão da MLP utilizando o conjunto de teste. Em seguida, identifique quais classes foram mais confundidas pelo modelo e discuta possíveis razões para esses erros de classificação, considerando as limitações de uma arquitetura MLP aplicada a problemas de visão computacional.</font>

```python
y_pred_prob = modelo_mlp.predict(x_test)
y_pred = np.argmax(y_pred_prob, axis=1)

matriz_confusao = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(
    matriz_confusao,
    annot=True,
    fmt='d',
    xticklabels=class_names,
    yticklabels=class_names,
    cmap='Blues'
)

plt.xlabel('Classe predita')
plt.ylabel('Classe verdadeira')
plt.title('Matriz de Confusão - MLP')
plt.show()
```

A diagonal principal mostra os acertos do modelo e os valores fora dela mostram as classes confundidas. A MLP costuma confundir classes parecidas, como gato e cachorro, carro e caminhão, ou pássaro e avião, porque ao usar Flatten ela perde a estrutura espacial das imagens. Sem essa informação, fica mais difícil identificar formas, bordas e texturas, que são justamente as características importantes em visão computacional.

### <font color='#2D9CDB'>Q14) Com base nos resultados obtidos ao longo desta etapa, discuta as principais vantagens e limitações das redes MLP para classificação de imagens. Em sua resposta, considere aspectos como preservação da estrutura espacial das imagens, quantidade de parâmetros treináveis, capacidade de generalização e desempenho obtido no CIFAR-10.</font>

A MLP é simples de criar e treinar, mas tem várias limitações para imagens. # <font color='green'><u><b>Parte 3 - CNN</b></u></font>

### <font color='#2D9CDB'>Q15) As redes neurais convolucionais (CNNs) foram desenvolvidas para lidar melhor com dados que possuem estrutura espacial, como imagens. Explique por que arquiteturas MLP apresentam limitações em problemas de visão computacional e descreva como as CNNs conseguem preservar relações espaciais entre os pixels. Em seguida, explique os conceitos de convolução, filtros convolucionais e mapas de características (feature maps).</font>

As MLPs têm limitações em visão computacional porque transformam a imagem em vetor com o Flatten, perdendo a relação espacial entre os pixels. Por isso elas não analisam bem padrões como bordas, formas e texturas. Já as CNNs trabalham diretamente com altura, largura e canais, mantendo a estrutura espacial. A convolução é o processo em que pequenos filtros percorrem a imagem procurando padrões locais. Esses filtros convolucionais aprendem características importantes como linhas, curvas e texturas, e o resultado desse processo são os feature maps, que mostram onde determinados padrões foram encontrados.

### <font color='#2D9CDB'>Q16) Construa uma rede neural convolucional utilizando a API Sequential do Keras. A arquitetura da CNN deve conter camadas Conv2D com ativação ReLU, camadas MaxPooling2D, uma camada Flatten e uma camada de saída com 10 neurônios e ativação Softmax. Após construir a arquitetura, explique o papel das camadas convolucionais, das camadas de pooling e da camada Flatten dentro da CNN.</font>

```python
modelo_cnn = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(10, activation='softmax')
])
modelo_cnn.summary()
```

As camadas Conv2D extraem características da imagem, como bordas, formas e texturas, preservando a relação entre os pixels. As camadas MaxPooling2D reduzem o tamanho dos mapas de características, mantendo a informação mais importante e diminuindo o custo computacional. A camada Flatten transforma o resultado da parte convolucional em um vetor, para que a camada densa final consiga gerar as probabilidades de cada uma das 10 classes através da Softmax.

### <font color='#2D9CDB'>Q17) Exiba o resumo da arquitetura da CNN utilizando o método summary() e compare a quantidade de parâmetros treináveis da CNN com a da MLP construída anteriormente. Em seguida, discuta por que CNNs costumam ser mais eficientes para classificação de imagens, mesmo utilizando menos parâmetros treináveis.</font>

```python
modelo_cnn.summary()
```

A CNN tem bem menos parâmetros que a MLP, mesmo aprendendo melhor as imagens. Isso acontece porque as camadas convolucionais usam filtros pequenos que são aplicados em várias regiões da imagem, em vez de conectar todos os pixels a todos os neurônios. Assim a CNN consegue ser mais eficiente e ainda aprender bem as características visuais.

### <font color='#2D9CDB'>Q18) Compile a CNN utilizando o otimizador Adam com taxa de aprendizado de $10^{-3}$, a função de custo sparse_categorical_crossentropy e a métrica accuracy. Em seguida, treine o modelo por até 25 épocas utilizando os conjuntos de treinamento e validação. Armazene o histórico do treinamento em um objeto do tipo History. Após o treinamento, plote gráficos mostrando a evolução da loss e da accuracy ao longo das épocas para os conjuntos de treinamento e validação e analise o comportamento do treinamento da CNN.</font>

```python
modelo_cnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

historico_cnn = modelo_cnn.fit(
    x_train,
    y_train,
    epochs=25,
    validation_data=(x_val, y_val)
)

plt.figure(figsize=(8, 5))
plt.plot(historico_cnn.history['loss'], label='Treino')
plt.plot(historico_cnn.history['val_loss'], label='Validação')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(historico_cnn.history['accuracy'], label='Treino')
plt.plot(historico_cnn.history['val_accuracy'], label='Validação')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

No treino da CNN espera-se que a loss diminua e a accuracy aumente tanto no treino quanto na validação. Como ela preserva melhor a estrutura espacial das imagens, tende a ter desempenho superior à MLP. Se a accuracy de treino continuar subindo muito enquanto a de validação para ou cai, é sinal de overfitting. Quando as duas curvas evoluem próximas, indica que o modelo está generalizando bem.

### <font color='#2D9CDB'>Q19) Avalie o desempenho da CNN utilizando o conjunto de teste e compare os resultados obtidos com aqueles alcançados pela MLP. Em seguida, discuta as diferenças observadas em termos de accuracy, capacidade de generalização e aprendizado de características visuais.</font>

```python
cnn_loss, cnn_accuracy = modelo_cnn.evaluate(x_test, y_test)

print(cnn_loss)
print(cnn_accuracy)

mlp_loss, mlp_accuracy = modelo_mlp.evaluate(x_test, y_test)
cnn_loss, cnn_accuracy = modelo_cnn.evaluate(x_test, y_test)

print(mlp_loss)
print(mlp_accuracy)

print(cnn_loss)
print(cnn_accuracy)
```

A CNN apresenta accuracy maior que a MLP no teste, pois consegue preservar a estrutura espacial das imagens e aprender características visuais como bordas, formas e texturas. A MLP usa o Flatten logo no início e perde essas relações entre os pixels, então tem mais dificuldade. Por isso a CNN costuma ter melhor generalização e ser mais adequada para classificação de imagens no CIFAR-10.

### <font color='#2D9CDB'>Q22) Adicione camadas Dropout à CNN e treine novamente o modelo. Em seguida, plote as curvas de loss e accuracy para os conjuntos de treinamento e validação e compare os resultados com a CNN original. Com base nos gráficos obtidos, discuta o impacto do Dropout no overfitting e na capacidade de generalização da rede neural.</font>

```python
modelo_cnn_dropout = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(10, activation='softmax')
])

modelo_cnn_dropout.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

historico_cnn_dropout = modelo_cnn_dropout.fit(
    x_train,
    y_train,
    epochs=25,
    validation_data=(x_val, y_val)
)

plt.figure(figsize=(8, 5))
plt.plot(historico_cnn_dropout.history['loss'], label='Treino com Dropout')
plt.plot(historico_cnn_dropout.history['val_loss'], label='Validação com Dropout')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(historico_cnn_dropout.history['accuracy'], label='Treino com Dropout')
plt.plot(historico_cnn_dropout.history['val_accuracy'], label='Validação com Dropout')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

O Dropout ajuda a reduzir o overfitting, pois desativa aleatoriamente parte dos neurônios durante o treinamento. Com isso a rede não fica tão dependente de caminhos específicos e aprende padrões mais gerais. Comparando com a CNN original, é comum a accuracy de treino ficar um pouco menor, mas a validação tende a ficar mais estável. Se a diferença entre treino e validação diminuir, isso mostra que o modelo generaliza melhor.
